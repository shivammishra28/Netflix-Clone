```python
POPULATION_SIZE = 10
CHROMOSOME_LENGTH = 5
GENERATIONS = 50
MUTATION_RATE = 0.01
```

```python
def decode(chromosome):
    return int("".join(map(str, chromosome)), 2)
```

```python
decode([0,0,0,1,1])
```

```python
def fitness(chromosome):
    #print("received chromo: ",chromosome)
    x = decode(chromosome)
    #print("decoded chromo: ",x)
    return x ** 2
```

```python
fitness([1,0,0,1,1])
```

```python
import random
def create_population():
    return [
        [random.randint(0, 1)
         for _ in range(CHROMOSOME_LENGTH)]
        for _ in range(POPULATION_SIZE)
    ]
```

```python
create_population()
```

```python
def selection(population):
    candidates = random.sample(
        population, min(3, len(population))
    )
    print(candidates)
    return max(candidates, key=fitness).copy()
```

```python
selection(create_population())
```

```python
def crossover(parent1, parent2):
    point = random.randint(1, CHROMOSOME_LENGTH - 1)

    child1 = parent1[:point] + parent2[point:]
    child2 = parent2[:point] + parent1[point:]

    return child1, child2
```

```python
def mutation(chromosome):
    for i in range(len(chromosome)):
        if random.random() < MUTATION_RATE:
            chromosome[i] = 1 - chromosome[i]
    return chromosome
```

```python
# Main genetic algorithm
def genetic_algorithm():
    population = create_population()

    for generation in range(GENERATIONS):
        population.sort(key=fitness, reverse=True)

        best = population[0]
        print(
            f"Generation {generation}: "
            f"x = {decode(best)}, "
            f"fitness = {fitness(best)}"
        )

        # Stop if the global optimum is found
        if decode(best) == 31:
            break

        # Preserve the best solution (elitism)
        new_population = [best.copy()]

        # Create the next generation
        while len(new_population) < POPULATION_SIZE:
            parent1 = selection(population)
            parent2 = selection(population)

            child1, child2 = crossover(parent1, parent2)

            new_population.append(mutation(child1))

            if len(new_population) < POPULATION_SIZE:
                new_population.append(mutation(child2))

        population = new_population

    best = max(population, key=fitness)

    print("\nBest solution found:")
    print("Chromosome:", "".join(map(str, best)))
    print("Value of x:", decode(best))
    print("Fitness:", fitness(best))

# Run the algorithm
if __name__ == "__main__":
    genetic_algorithm()
```
