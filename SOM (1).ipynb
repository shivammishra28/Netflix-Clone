```python
import numpy as np
```

```python
class SOM:
    def __init__(self, m, n, dim, lr=0.5, n_iter=100):
        self.m, self.n, self.dim = m, n, dim
        self.lr = lr
        self.n_iter = n_iter
        # Initialize weights randomly
        self.weights = np.random.rand(m, n, dim)

    def _find_bmu(self, x):
        """Best Matching Unit (BMU) = neuron closest to input x"""
        diff = self.weights - x
        dist=np.sqrt(np.sum(diff**2, axis=2)) 

        #dist = np.linalg.norm(diff, axis=2)
        return np.unravel_index(np.argmin(dist), (self.m, self.n))

    def train(self, data):
        for t in range(self.n_iter):
            # pick one sample randomly
            x = data[np.random.randint(len(data))]
            bmu = self._find_bmu(x)
            i, j = bmu
            # update only the BMU (since only 2 clusters)
            self.weights[i, j] += self.lr * (x - self.weights[i, j])

    def map(self, data):
        return [self._find_bmu(x) for x in data]
```

```python
# --- Tiny dataset: 4 samples, 2 features ---
# First two form cluster1, last two form cluster2
X = np.array([
    [1.0, 1.5],
    [1.2, 1.3],
    [4.0, 4.5],
    [3.8, 4.2]
])

# --- Train a 1x2 SOM (two nodes) ---
som = SOM(m=1, n=2, dim=2, n_iter=200)
som.train(X)

# --- Map samples to nodes ---
mapped = som.map(X)

print("Final SOM Weights (2 clusters):")
print(som.weights.reshape(2,2))

print("\nSample -> Cluster assignment:")
for i, x in enumerate(X):
    print(f"Sample {x} -> Cluster {mapped[i][1]}")  # use col index (0 or 1)
```
