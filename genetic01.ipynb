Genetic Algorithm to minimize f(x)=a+2b+3c+4d-30

```
import random
def population(chromosome, genes):
  pop=[]
  for i in range(chromosome):
    pop.append([])
    for j in range(genes):
      pop[i].append(random.randint(0,30))
  return pop
```

```
pop=population(6,4)
```

```
pop
```

```
def fitness(pop):
  result1=[]
  for i in range(len(pop)):
    result=pop[i][0]+pop[i][1]*2+pop[i][2]*3+pop[i][3]*4-30
    #result=1/(1+result)
    print(result)
    result1.append(result)
  return(result1)
```

```
fit=fitness(pop)
```

```
total_fit=sum(fit)
```

```
total_fit
```

```
def fit_probability(fit, total):
  probability_fit=[]
  for i in range(len(fit)):
    probability_fit.append(fit[i]/total)
  return probability_fit
```

```
probability_fit=fit_probability(fit, total_fit)
```

```
probability_fit
```

```
def comm_prob(fitn):
  comm_fitness=[]
  s=0
  for i in range(len(fitn)):
    s=s+fitn[i]
    comm_fitness.append(s)
  return comm_fitness
```

```
comm_fitness=comm_prob(probability_fit)
```

```
comm_fitness
```

```
def roulette_wheel(pop, comm_fitness):
  new_pop=[]
  r=[]
  for i in range(6):
    r.append(random.random())
  print(r)
  print(comm_fitness)
  for i in range(6):
    for j in range(6):
      if r[i]<=comm_fitness[j]:
        new_pop.append(pop[j])
        break

  return new_pop
```

```
new_pop=roulette_wheel(pop, comm_fitness)
```

```
new_pop
```

```
def tournament(fit, pop):
  new_pop_t=[]
  for i in range(len(pop)):
    r1=random.randint(0,5)
    r2=random.randint(0,5)
    if fit[r1]>fit[r2]:
      new_pop_t.append(pop[r1])
    else:
      new_pop_t.append(pop[r2])
  return new_pop_t
```

```
new_pop_t=tournament(fit, pop)
```

```
new_pop_t
```

```
def crossover(cross_rate, pop):
  r=[]
  r1=[]
  for i in range(len(pop)):
    r.append(random.random())
  chromo_cross=[]
  for i in range(len(pop)):
    if r[i]<cross_rate:
      chromo_cross.append(i)
  print(chromo_cross)
  for i in range(len(chromo_cross)):
    r1.append(random.randint(1,2))
  for i in range(len(chromo_cross)):
    if i==len(chromo_cross)-1:
      pop[chromo_cross[i]]=pop[chromo_cross[i]][:r1[i]]+pop[chromo_cross[0]][r1[i]:]
    else:
      pop[chromo_cross[i]]=pop[chromo_cross[i]][:r1[i]]+pop[chromo_cross[i+1]][r1[i]:]
  return pop
```

```
new_pop=crossover(0.7, new_pop)
```

```
new_pop
```

```
def mutation(mrate, pop):
  chr=len(pop)
  genes=len(pop[0])
  total_genes=chr*genes
  mutate_genes=int(total_genes*mrate)
  for i in range(mutate_genes):
    r=random.randint(0,total_genes-1)
    row=r//genes
    col=r%genes
    value=random.randint(0,30)
    pop[row][col]=value
  return pop
```

```
mutated_pop=mutation(0.2, new_pop)
```

```
mutated_pop
```

```
chromo=6
gene=4
initial_pop=population(chromo, gene)
for i in range(10):
  print("Iteration: ",i)
  fit_values=fitness(initial_pop)
  print(fit_values)
  new_pop=tournament(fit_values, initial_pop)
  cross_pop=crossover(0.7, new_pop)
  mutated_pop=mutation(0.2, cross_pop)
  initial_pop=mutated_pop
```
