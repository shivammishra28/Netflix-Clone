```python
# Define fuzzy variables
service = ctrl.Antecedent(np.arange(0, 11, 1), 'service')
food = ctrl.Antecedent(np.arange(0, 11, 1), 'food')
tip = ctrl.Consequent(np.arange(0, 26, 1), 'tip')
```

```python

# Define membership functions for service
service['poor'] = fuzz.trimf(service.universe, [0, 0, 3])
service['fair'] = fuzz.trimf(service.universe, [2, 4, 6])
service['good'] = fuzz.trimf(service.universe, [5, 7, 9])
service['excellent'] = fuzz.trimf(service.universe, [7, 10, 10])
```

```python

# Define membership functions for food
food['bad'] = fuzz.trimf(food.universe, [0, 0, 4])
food['average'] = fuzz.trimf(food.universe, [3, 5, 7])
food['delicious'] = fuzz.trimf(food.universe, [6, 10, 10])
```

```python
# Define membership functions for tip
tip['very_low'] = fuzz.trimf(tip.universe, [0, 0, 5])
tip['low'] = fuzz.trimf(tip.universe, [0, 5, 10])
tip['medium'] = fuzz.trimf(tip.universe, [5, 12, 18])
tip['high'] = fuzz.trimf(tip.universe, [15, 20, 23])
tip['very_high'] = fuzz.trimf(tip.universe, [20, 25, 25])
```

```python
# Define fuzzy rules
rule1 = ctrl.Rule(service['poor'] & food['bad'], tip['very_low'])
rule2 = ctrl.Rule(service['poor'] & food['average'], tip['low'])
rule3 = ctrl.Rule(service['fair'] & food['bad'], tip['low'])
rule4 = ctrl.Rule(service['fair'] & food['average'], tip['medium'])
rule5 = ctrl.Rule(service['good'] & food['average'], tip['high'])
rule6 = ctrl.Rule(service['good'] & food['delicious'], tip['very_high'])
rule7 = ctrl.Rule(service['excellent'] & food['delicious'], tip['very_high'])
rule8 = ctrl.Rule(service['excellent'] & food['average'], tip['high'])
rule9 = ctrl.Rule(service['poor'] & food['delicious'], tip['medium'])
```

```python
# Create control system
tipping_ctrl = ctrl.ControlSystem([rule1, rule2, rule3, rule4, rule5, rule6,
                                   rule7, rule8, rule9])
tipping = ctrl.ControlSystemSimulation(tipping_ctrl)
```

```python

# Input values
tipping.input['service'] = 9.5
tipping.input['food'] = 7.5

# Compute output
tipping.compute()

# Display results
print(f"Calculated Tip: {np.floor(tipping.output['tip'])}")

# (Optional) Visualize
# tip.view(sim=tipping)
```
